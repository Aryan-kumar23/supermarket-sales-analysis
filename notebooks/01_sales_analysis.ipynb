# 2. Data Cleaning and Preprocessing

The purpose of this stage is to improve data quality and
prepare the supermarket sales dataset for exploratory
data analysis.

The following activities will be performed:

1. Create a copy of the original dataset.
2. Check column names and data consistency.
3. Handle missing values if necessary.
4. Identify duplicate records.
5. Correct data types.
6. Validate numerical values.
7. Verify the cleaned dataset.

```python
# Create a copy of the original dataset

df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)

print("\nDataset copied successfully!")
```

```python
# Remove unnecessary spaces from column names

df_clean.columns = df_clean.columns.str.strip()

# Remove leading and trailing spaces from text columns

for column in df_clean.select_dtypes(include="object").columns:
    df_clean[column] = df_clean[column].str.strip()

print("Column names and text values cleaned successfully!")

print(df_clean.columns.tolist())
```

```python
# Calculate missing values

missing_values = df_clean.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": (
        missing_values / len(df_clean) * 100
    ).round(2)
})

display(missing_report)
```

```python
# Count completely duplicated rows

duplicate_count = df_clean.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
```

```python
# Display duplicate records, if any

duplicates = df_clean[
    df_clean.duplicated(keep=False)
]

display(duplicates.head(20))
```

```python
print(df_clean.dtypes)
```

```python
# Convert Date column into datetime format

df_clean["Date"] = pd.to_datetime(
    df_clean["Date"],
    format="%m/%d/%Y",
    errors="coerce"
)

print("Date conversion completed!")

display(df_clean[["Date"]].head())
```

```python
# Numerical columns

numeric_columns = [
    "Unit price",
    "Quantity",
    "Tax 5%",
    "Total",
    "cogs",
    "gross margin percentage",
    "gross income",
    "Rating"
]

# Convert numerical columns

for column in numeric_columns:
    if column in df_clean.columns:
        df_clean[column] = pd.to_numeric(
            df_clean[column],
            errors="coerce"
        )

print("Numerical conversion completed!")
```

```python
# Check invalid quantities

invalid_quantity = df_clean[
    df_clean["Quantity"].notna() &
    (df_clean["Quantity"] <= 0)
]

# Check invalid unit prices

invalid_price = df_clean[
    df_clean["Unit price"].notna() &
    (df_clean["Unit price"] <= 0)
]

# Check invalid customer ratings

invalid_rating = df_clean[
    df_clean["Rating"].notna() &
    (~df_clean["Rating"].between(0, 10))
]

print("Invalid quantities:", len(invalid_quantity))
print("Invalid unit prices:", len(invalid_price))
print("Invalid customer ratings:", len(invalid_rating))
```

```python
# Calculate expected total

expected_total = (
    df_clean["cogs"] + df_clean["Tax 5%"]
)

# Check rows with valid numerical values

valid_rows = df_clean[
    ["Sales", "cogs", "Tax 5%"]
].notna().all(axis=1)

# Compare expected and actual total

mismatch = valid_rows & ~np.isclose(
    df_clean["Sales"],
    expected_total,
    rtol=0.0001,
    atol=0.01
)

print("Total calculation mismatches:", mismatch.sum())
```

# 3. Sales Analysis and Business Insights

In this section, we analyze supermarket sales data to identify
important business patterns.

### Business Questions

1. What are the total sales and average transaction value?
2. Which product line generates the highest sales?
3. Which branch performs best?
4. Which payment method is most popular?
5. Which customer type contributes more sales?
6. How do sales change over time?
7. Which day of the week has the highest sales?

```python
# Key Performance Indicators (KPIs)

total_sales = df_clean["Sales"].sum()

total_transactions = df_clean["Invoice ID"].nunique()

total_quantity = df_clean["Quantity"].sum()

average_transaction = df_clean["Sales"].mean()

total_gross_income = df_clean["gross income"].sum()

average_rating = df_clean["Rating"].mean()

# Display KPIs

kpi_summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Transactions",
        "Total Quantity Sold",
        "Average Transaction Value",
        "Total Gross Income",
        "Average Customer Rating"
    ],
    "Value": [
        total_sales,
        total_transactions,
        total_quantity,
        average_transaction,
        total_gross_income,
        average_rating
    ]
})

display(kpi_summary)
```

```python
# Sales by product line

product_analysis = (
    df_clean.groupby("Product line")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Average_Sales=("Sales", "mean"),
        Gross_Income=("gross income", "sum")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

display(product_analysis)
```

```python
plt.figure(figsize=(11, 6))

sns.barplot(
    data=product_analysis.reset_index(),
    x="Total_Sales",
    y="Product line"
)

plt.title("Total Sales by Product Line")
plt.xlabel("Total Sales")
plt.ylabel("Product Category")

plt.tight_layout()
plt.show()
```

```python
branch_analysis = (
    df_clean.groupby("Branch")
    .agg(
        Total_Sales=("Sales", "sum"),
        Transactions=("Invoice ID", "nunique"),
        Quantity_Sold=("Quantity", "sum"),
        Gross_Income=("gross income", "sum")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

display(branch_analysis)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=branch_analysis.reset_index(),
    x="Branch",
    y="Total_Sales"
)

plt.title("Sales Performance by Branch")
plt.xlabel("Branch")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
payment_analysis = (
    df_clean.groupby("Payment")
    .agg(
        Transactions=("Invoice ID", "nunique"),
        Total_Sales=("Sales", "sum")
    )
    .sort_values(
        by="Transactions",
        ascending=False
    )
)

display(payment_analysis)
```

```python
plt.figure(figsize=(7, 7))

payment_analysis["Transactions"].plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Payment Method Distribution")
plt.ylabel("")

plt.tight_layout()
plt.show()
```

```python
customer_analysis = (
    df_clean.groupby("Customer type")
    .agg(
        Total_Sales=("Sales", "sum"),
        Transactions=("Invoice ID", "nunique"),
        Average_Sales=("Sales", "mean"),
        Gross_Income=("gross income", "sum")
    )
)

display(customer_analysis)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=customer_analysis.reset_index(),
    x="Customer type",
    y="Total_Sales"
)

plt.title("Sales by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
# Ensure Date is datetime

df_clean["Date"] = pd.to_datetime(
    df_clean["Date"],
    errors="coerce"
)

# Prepare date-based data

date_data = df_clean.dropna(
    subset=["Date"]
).copy()

# Calculate monthly sales

date_data["Month"] = (
    date_data["Date"].dt.to_period("M").astype(str)
)

monthly_sales = (
    date_data.groupby("Month")["Sales"]
    .sum()
)

display(monthly_sales)
```

```python
plt.figure(figsize=(10, 5))

monthly_sales.plot(
    kind="line",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
date_data["Day"] = date_data["Date"].dt.day_name()

days_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_sales = (
    date_data.groupby("Day")["Sales"]
    .sum()
    .reindex(days_order)
)

display(day_sales)
```

```python
plt.figure(figsize=(11, 5))

day_sales.plot(kind="bar")

plt.title("Sales by Day of the Week")
plt.xlabel("Day")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
